#**Data Wrangling**

```python
import pandas as pd

students= pd.DataFrame({
    'Student_ID':[101,102,103,104],
    'Name':['Aman','Priya','Rahul','Shagun'],
    'Dept':["CSE","CSE","DS","ECE"],
    'Marks':[85,90,78,95]
})

marks=pd.DataFrame({
    'Student_ID':[101,102,104,105],
    'Marks':[85,90,78,95]
})

print("Students Table:")
print(students)
print("\nMarks Table:")
print(marks)
```

```python
inner=pd.merge(students,marks,on='Student_ID')
print("\nInner Join Result:")
print(inner)
```

```python
left=pd.merge(students,marks,on='Student_ID',how='left')
print("\nLeft Join Result:")
print(left)
```

```python
right=pd.merge(students,marks,on='Student_ID',how='right')
print("\nRight Join Result:")
print(right)
```

```python
outer=pd.merge(students,marks,on='Student_ID',how='outer')
print("\nOuter Join Result:")
print(outer)
34235
```

```python
students_info=students.rename(columns={'Student_ID':'ID'})
print("\nStudents Info Table:")
print(students_info)

combined=pd.merge(students_info,marks,left_on='ID',right_on='Student_ID',how = 'inner')
print("\nCombined Result:")
print(combined)
```

***Join using an Index.***

```python
students_indexed=students.set_index('Student_ID')
marks_indexed=marks.set_index('Student_ID')

joined = students_indexed.join(marks_indexed, how='left', lsuffix='_left', rsuffix='_right')
print("\nStudents Indexed:")
print(students_indexed)
print("\nMarks Indexed:")
print(marks_indexed)
print("\nJoined Result:")
print(joined)
```

```python

import pandas as pd
import numpy as np
raw_students = pd.DataFrame({
    "Roll_No": [1, 2, 3, 3, 4, 5],
    "Name": [" aman ", "PRIYA", "Rahul", "Rahul", "Neha", "Isha"],
    "Department": ["cse", "IT", "CSE", "CSE", "ece", "IT"],
    "Marks": ["85", "90", "78", "78", "missing", "67"],
    "Attendance": [92, 95, 80, 80, 88, np.nan]
})
print("Original Data:")
print(raw_students)
print("\nShape:")
print(raw_students.shape)
print("\nData Types:")
print(raw_students.dtypes)
print("\nMissing Values:")
print(raw_students.isnull().sum())
print("\nDuplicate Records:")
students = raw_students.copy()
students["Name"] = students["Name"].str.strip().str.title()
students["Department"] = students["Department"].str.strip().str.upper()
students["Marks"] = pd.to_numeric(
    students["Marks"].replace("missing", np.nan),
    errors="coerce"
)
students = students.drop_duplicates(subset="Roll_No", keep="first")
students["Marks"] = students["Marks"].fillna(
    students["Marks"].mean()
)
students["Attendance"] = students["Attendance"].fillna(
    students["Attendance"].mean()
)
students["Pass"] = students["Marks"] >= 40
department_avg = students.groupby("Department")[
    ["Marks", "Attendance"]
].mean()
print("\nCleaned Data:")
print(students)
print("\nDepartment-wise Average:")
print(department_avg)
students.to_csv("cleaned_students.csv", index=False)
students.to_pickle("cleaned_students.pkl")
print("\nFiles saved successfully!")
```

```python

import pandas as pd
# Q1.
df1 = pd.DataFrame({"ID": [1, 2, 3], "Name": ["Aman", "Priya", "Rahul"]})
df2 = pd.DataFrame({"ID": [2, 3, 4], "Marks": [80, 90, 85]})
result = pd.merge(df1, df2, on="ID", how="outer", indicator=True)
print("Outer Merge:")
print(result)
print("Non-matching IDs:", result.loc[result["_merge"] != "both", "ID"].tolist())
# Q2.
df3 = pd.DataFrame({"ID": [1, 2, 3], "Name": ["Aman", "Priya", "Rahul"]})
df4 = pd.DataFrame({"ID": [1, 2, 3], "Marks": [85, 90, 78]})
result2 = pd.merge(df3, df4, on="ID", validate="one_to_one")
print("\nOne-to-One Merge:")
print(result2)
# Q3.
print("\nUsing merge():")
print(pd.merge(df3, df4, on="ID"))
print("\nUsing join():")
print(df3.set_index("ID").join(df4.set_index("ID")))
# Q4.
df5 = pd.DataFrame({"ID": [1, 1], "Name": ["Aman", "Rahul"]})
df6 = pd.DataFrame({"ID": [1, 1], "Marks": [80, 90]})
result3 = pd.merge(df5, df6, on="ID")
print("\nMany-to-Many Merge:")
print(result3)
print("Total rows:", len(result3))
# Q5.
df7 = pd.DataFrame({
    "Student_ID": [101, 102, 101],
    "Semester": [1, 1, 2],
    "Name": ["Aman", "Priya", "Aman"]
})
df8 = pd.DataFrame({
    "Student_ID": [101, 102, 101],
    "Semester": [1, 1, 2],
    "Marks": [85, 90, 88]
})
result4 = pd.merge(
    df7, df8,
    on=["Student_ID", "Semester"],
    how="inner"
)
print("\nMerge on Two Columns:")
print(result4)
```
